## 🟡 Étape 1 — Les variables

### 📖 Rappel
Une **variable** stocke une valeur. On peut la changer à tout moment.
```python
vitesse = 3        # int → nombre entier
couleur = "red"    # str → texte entre guillemets
```

Dans le casse-brique, les variables **règlent** le jeu :

| Variable | Effet |
|----------|-------|
| `vitesse_balle = 3` | Vitesse — essaie `2`, `4`, `6`... |
| `raquette_largeur = 80` | Taille de la raquette en pixels |
| `couleur_brique = "red"` | Couleur des briques |

### ✏️ Ton défi
Dans la cellule du jeu, zone `ÉTAPE 1` :
- Choisis une couleur pour les briques (`"red"`, `"blue"`, `"#FF6B6B"`...)
- Choisis une couleur pour la raquette
- Essaie différentes vitesses !

## 🔵 Étape 2 — La liste

### 📖 Rappel
Une **liste** stocke plusieurs valeurs dans l'ordre.
```python
positions = [50, 150, 250, 350]  # 4 nombres = 4 positions
positions[0]    # → 50  (premier élément, indice 0)
len(positions)  # → 4   (nombre d'éléments)
```

Dans le casse-brique, chaque **nombre** = la position X d'une brique :
```
écran :  0─────────────────────────────480
briques :  [50]    [150]    [250]    [350]
```
➡️ Plus tu mets de nombres, plus il y a de briques !

### ✏️ Ton défi
Dans la zone `ÉTAPE 2`, remplace les `___` par des nombres entre **0 et 430**.
> 💡 Pour espacer les briques : ajoute environ **100** entre chaque nombre
> Exemple : `[20, 120, 220, 320]` → 4 briques bien espacées

## 🟣 Étape 3 — Les conditions

### 📖 Rappel
Un **`if`** permet de réagir à une situation.
```python
if balle_x + rayon >= 480:    # si la balle touche le bord droit
    vitesse_x = -vitesse_x    # elle repart dans l'autre sens
```

La balle rebondit parce que sa vitesse **change de signe** :
- `vitesse_x = +3` → va **à droite**
- `vitesse_x = -3` → va **à gauche**
- `vitesse_y = +3` → va **vers le bas**
- `vitesse_y = -3` → va **vers le haut**

Pour forcer un sens précis, on utilise `abs()` (valeur absolue = toujours positive) :
```python
vitesse_x = abs(vitesse_x)   # → toujours vers la droite
vitesse_y = abs(vitesse_y)   # → toujours vers le bas
```

### ✏️ Ton défi
Dans la zone `ÉTAPE 3`, réponds aux deux questions :
- Mur **gauche** → la balle doit repartir vers la **droite** (vitesse positive)
- **Plafond** → la balle doit repartir vers le **bas** (vitesse positive)

> 💡 **Indice :** `abs(vitesse_balle)` donne toujours un nombre positif !

````python
from IPython.display import display, HTML
import json

# ════════════════════════════════════════════════════════════
#  ✏️ ÉTAPE 1 — Les variables : règle ton jeu !
# ════════════════════════════════════════════════════════════

vitesse_balle    = 2        # ← Essaie 2, 4, 5...
raquette_largeur = 80       # ← Essaie 50 (difficile) ou 120 (facile)
couleur_brique   = "red"    # ← "red" / "blue" / "green" / "orange" / "#FF6B6B"
couleur_balle    = "white"  # ← Couleur de la balle
couleur_raquette = "blue"    # ← Essaie "#42a5f5" ou "yellow" !

# ════════════════════════════════════════════════════════════
#  ✏️ ÉTAPE 2 — La liste : place tes briques !
#  Chaque nombre = position X d'une brique (entre 0 et 430)
#  Exemple : [20, 120, 220, 320]  → 4 briques bien espacées
# ════════════════════════════════════════════════════════════

positions_briques = [20 , 160]   # ← remplace les ___ !

# ════════════════════════════════════════════════════════════
#  ✏️ ÉTAPE 3 — Les conditions : fais rebondir la balle !
#
#  Le jeu gère déjà le rebond sur le mur DROIT :
#    if balle_x + rayon >= 480:        ← balle touche le bord droit
#        vitesse_x = -abs(vitesse_x)   ← repart VERS LA GAUCHE (négatif)
#
#  À toi de répondre :
# ════════════════════════════════════════════════════════════

# Mur GAUCHE → balle doit repartir VERS LA DROITE (vitesse positive)
rebond_gauche_vitesse = abs(vitesse_balle)    # ← abs(vitesse_balle)  ou  -vitesse_balle ?

# PLAFOND → balle doit repartir VERS LE BAS (vitesse positive)
rebond_plafond_vitesse = abs(vitesse_balle)   # ← abs(vitesse_balle)  ou  -vitesse_balle ?

# ════════════════════════════════════════════════════════════
#  🔒 Vérifications et lancement — ne pas modifier
# ════════════════════════════════════════════════════════════
erreurs = []
if couleur_brique == "___":
    erreurs.append('couleur_brique : remplace ___ — ex: "red"')
if couleur_raquette == "___":
    erreurs.append('couleur_raquette : remplace ___ — ex: "#42a5f5"')
try:
    if any(str(p) == '___' for p in positions_briques):
        erreurs.append('positions_briques : remplace les ___ par des nombres — ex: 20, 120, 220, 320')
except TypeError:
    erreurs.append('positions_briques : doit être une liste — ex: [20, 120, 220, 320]')
try:
    if str(rebond_gauche_vitesse) == '___' or str(rebond_plafond_vitesse) == '___':
        erreurs.append('Complète rebond_gauche_vitesse et rebond_plafond_vitesse !')
except: pass

if erreurs:
    print('⚠️  Corrige ces erreurs avant de lancer le jeu :')
    for e in erreurs: print(f'  ❌ {e}')
else:
    print(f'✅ {len(positions_briques)} briques | vitesse={vitesse_balle} | raquette={raquette_largeur}px')
    print('🎮 Le jeu se lance ci-dessous !')
    config = {
        'vitesse_x': vitesse_balle, 'vitesse_y': vitesse_balle,
        'raquette_largeur': raquette_largeur, 'balle_rayon': 8,
        'couleur_brique': couleur_brique, 'couleur_balle': couleur_balle,
        'couleur_raquette': couleur_raquette, 'positions_briques': positions_briques,
        'brique_largeur': 80, 'brique_y': 60,
        'rebond_gauche': rebond_gauche_vitesse,
        'rebond_plafond': rebond_plafond_vitesse,
    }
    cfg_json = json.dumps(config)
    GAME_HTML = "<!DOCTYPE html>\n<html lang=\"fr\">\n<head>\n<meta charset=\"UTF-8\">\n<style>\n  * { margin:0; padding:0; box-sizing:border-box; }\n  body { background:#0f0f1a; display:flex; flex-direction:column;\n         align-items:center; font-family:'Segoe UI',sans-serif; color:#fff; padding:10px; }\n  #wrapper { position:relative; }\n  canvas { display:block; border:2px solid #333; border-radius:8px; background:#1a1a2e; }\n  #ui { display:flex; justify-content:space-between; align-items:center;\n        padding:8px 4px; width:100%; }\n  .uv { font-size:16px; font-weight:700; color:#FFD600; }\n  .ul { font-size:11px; color:#888; text-transform:uppercase; letter-spacing:1px; }\n  .ub { text-align:center; }\n  #msg { position:absolute; top:50%; left:50%; transform:translate(-50%,-50%);\n         background:rgba(0,0,0,0.88); border:2px solid #FFD600; border-radius:14px;\n         padding:24px 36px; text-align:center; display:none; z-index:10; }\n  #msg h2 { font-size:22px; color:#FFD600; margin-bottom:8px; }\n  #msg p  { font-size:14px; color:#ccc; margin-bottom:16px; }\n  #btn-r { background:linear-gradient(135deg,#FFD600,#ff8f00); color:#1a1200;\n            border:none; border-radius:8px; padding:10px 24px; font-size:14px;\n            font-weight:900; cursor:pointer; }\n  #btn-r:hover { opacity:0.85; }\n  #toolbar { display:flex; justify-content:center; margin-top:8px; }\n  #btn-restart { background:linear-gradient(135deg,#FFD600,#ff8f00); color:#1a1200;\n                 border:none; border-radius:8px; padding:9px 28px; font-size:14px;\n                 font-weight:900; cursor:pointer; letter-spacing:0.5px; }\n  #btn-restart:hover { opacity:0.85; }\n  #btn-restart:active { transform:scale(0.97); }\n  #cpanel { margin-top:10px; background:#0d1117; border-radius:10px;\n             padding:12px 16px; width:100%; font-family:monospace; font-size:12px; }\n  .cpl { color:#888; font-size:10px; margin-bottom:6px; text-transform:uppercase; letter-spacing:1px; }\n  .cv{color:#ff79c6;} .cn{color:#bd93f9;} .cs{color:#50fa7b;} .cc{color:#6272a4;}\n</style>\n</head>\n<body>\n<div id=\"wrapper\">\n  <div id=\"ui\">\n    <div class=\"ub\"><div class=\"uv\" id=\"s\">0</div><div class=\"ul\">Score</div></div>\n    <div class=\"ub\"><div class=\"uv\" id=\"b\">0</div><div class=\"ul\">Briques</div></div>\n    <div class=\"ub\"><div class=\"uv\" id=\"v\">3</div><div class=\"ul\">Vies</div></div>\n  </div>\n  <canvas id=\"c\" width=\"480\" height=\"340\"></canvas>\n  <div id=\"msg\">\n    <h2 id=\"mt\">Gagn\u00e9 !</h2>\n    <p  id=\"mp\">Bravo !</p>\n    <button id=\"btn-r\" onclick=\"init()\">\ud83d\udd04 Rejouer</button>\n  </div>\n</div>\n<div id=\"toolbar\">\n  <button id=\"btn-restart\" onclick=\"init()\">\ud83d\udd04 Recommencer</button>\n</div>\n<div id=\"cpanel\">\n  <div class=\"cpl\">\ud83d\udd0d Valeurs actuelles dans le jeu</div>\n  <div id=\"cl\"></div>\n</div>\n<script>\nconst C=window.__CFG__;\nconst W=480,H=340,cv=document.getElementById('c'),ctx=cv.getContext('2d');\nlet bx,by,vx,vy,rx,briques,score,vies,run,aid;\nconst RW=C.raquette_largeur,RH=12,BR=C.balle_rayon;\n\nfunction init(){\n  score=0;vies=3;run=true;\n  document.getElementById('msg').style.display='none';\n  bx=W/2;by=H-80;vx=C.vitesse_x;vy=-C.vitesse_y;\n  rx=(W-RW)/2;\n  briques=C.positions_briques.map(x=>({x,y:C.brique_y,w:C.brique_largeur,h:18,ok:true}));\n  maj();if(aid)cancelAnimationFrame(aid);loop();\n}\n\nfunction loop(){\n  if(!run)return;\n  ctx.clearRect(0,0,W,H);\n  move();draw_briques();draw_raq();draw_ball();check_win();maj();\n  aid=requestAnimationFrame(loop);\n}\n\nfunction move(){\n  bx+=vx;by+=vy;\n  // Rebond mur droit\n  if(bx+BR>=W) vx=-Math.abs(vx);\n  // Rebond mur gauche \u2014 utilise C.rebond_gauche\n  if(bx-BR<=0) vx=C.rebond_gauche;\n  // Rebond plafond \u2014 utilise C.rebond_plafond\n  if(by-BR<=0) vy=C.rebond_plafond;\n  // Rebond raquette\n  if(by+BR>=H-RH-10&&by+BR<=H-5&&bx>=rx&&bx<=rx+RW){\n    vy=-Math.abs(vy);\n    let h=(bx-rx)/RW;\n    vx=C.vitesse_x*(h<0.5?-1:1)*(0.5+Math.abs(h-0.5)*1.5);\n  }\n  // Balle perdue\n  if(by>H){\n    vies--;\n    if(vies<=0)fin(false);\n    else{bx=W/2;by=H-80;vx=C.vitesse_x;vy=-C.vitesse_y;}\n  }\n  // Collisions briques\n  briques.forEach(b=>{\n    if(!b.ok)return;\n    if(bx+BR>b.x&&bx-BR<b.x+b.w&&by+BR>b.y&&by-BR<b.y+b.h){\n      b.ok=false;vy=-vy;score+=10;\n    }\n  });\n}\n\nfunction draw_ball(){\n  ctx.beginPath();ctx.arc(bx,by,BR,0,Math.PI*2);\n  ctx.fillStyle=C.couleur_balle;ctx.fill();\n}\nfunction draw_raq(){\n  ctx.fillStyle=C.couleur_raquette;\n  ctx.beginPath();ctx.roundRect(rx,H-RH-10,RW,RH,4);ctx.fill();\n}\nfunction draw_briques(){\n  briques.forEach(b=>{\n    if(!b.ok)return;\n    ctx.fillStyle=C.couleur_brique;\n    ctx.beginPath();ctx.roundRect(b.x,b.y,b.w,b.h,3);ctx.fill();\n    ctx.fillStyle='rgba(255,255,255,0.15)';\n    ctx.fillRect(b.x+3,b.y+3,b.w-6,4);\n  });\n}\nfunction check_win(){if(briques.every(b=>!b.ok))fin(true);}\nfunction fin(w){\n  run=false;\n  document.getElementById('mt').textContent=w?'\ud83c\udfc6 Gagn\u00e9 !':'\ud83d\udc80 Perdu !';\n  document.getElementById('mp').textContent=w?`Score : ${score} pts !`:`${score} pts \u2014 r\u00e9essaie !`;\n  document.getElementById('msg').style.display='block';\n}\nfunction maj(){\n  document.getElementById('s').textContent=score;\n  document.getElementById('b').textContent=briques.filter(b=>b.ok).length;\n  document.getElementById('v').textContent=vies;\n}\n\nlet keys={};\ndocument.addEventListener('keydown',e=>{keys[e.key]=true;e.preventDefault();});\ndocument.addEventListener('keyup',e=>{keys[e.key]=false;});\ncv.addEventListener('mousemove',e=>{\n  const r=cv.getBoundingClientRect();\n  rx=Math.max(0,Math.min(W-RW,e.clientX-r.left-RW/2));\n});\nsetInterval(()=>{\n  if(!run)return;\n  if(keys['ArrowLeft']) rx=Math.max(0,rx-6);\n  if(keys['ArrowRight'])rx=Math.min(W-RW,rx+6);\n},16);\n\n// Panel code live\ndocument.getElementById('cl').innerHTML=\n  `<div><span class=\"cv\">vitesse_balle</span> = <span class=\"cn\">${C.vitesse_x}</span> <span class=\"cc\">\u2190 plus grand = plus rapide</span></div>`+\n  `<div><span class=\"cv\">raquette_largeur</span> = <span class=\"cn\">${C.raquette_largeur}</span></div>`+\n  `<div><span class=\"cv\">couleur_brique</span> = <span class=\"cs\">\"${C.couleur_brique}\"</span></div>`+\n  `<div><span class=\"cv\">positions_briques</span> = <span class=\"cn\">[${C.positions_briques}]</span> <span class=\"cc\">\u2190 ${C.positions_briques.length} briques</span></div>`;\n\ninit();\n</script>\n</body>\n</html>"
    final = GAME_HTML.replace('const C=window.__CFG__;', f'const C={cfg_json};')
    display(HTML(f'<div style="max-width:520px;margin:0 auto;">{final}</div>'))
````
