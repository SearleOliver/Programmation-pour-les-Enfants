## 🟠 Étape 1 — Le dictionnaire : la fiche d'une brique

### 📖 Rappel
Un **dictionnaire** regroupe plusieurs infos sous un même nom.
```python
brique = {"x": 50, "y": 80, "couleur": "red", "points": 10}
brique["x"]       # → 50
brique["couleur"] # → "red"
```

### 🔍 Dans le jeu, une brique ressemble à ça :
```python
{
    "x":         50,      # position horizontale (pixels)
    "y":         80,      # position verticale (pixels)
    "w":         80,      # largeur (pixels)
    "h":         20,      # hauteur (pixels)
    "couleur":   "red",   # couleur affichée
    "points":    10,      # points gagnés quand on la casse
    "resistance": 1,      # nombre de coups pour la détruire
}
```

### ✏️ Ton défi — Exercice 1
Exécute la cellule ci-dessous et complète les `___` pour créer **une brique** valide.
Lis les commentaires — ils t'indiquent exactement quoi mettre !

````python
# ════════════════════════════════════════════════════════════
#  EXERCICE 1 — Crée le dictionnaire d'une brique
# ════════════════════════════════════════════════════════════

ma_brique = {
    "x":         50,      # position horizontale (pixels)
    "y":         80,      # position verticale (pixels)
    "w":         80,      # largeur (pixels)
    "h":         20,      # hauteur (pixels)
    "couleur":   "red",   # couleur affichée
    "points":    10,      # points gagnés quand on la casse
    "resistance": 1,      # nombre de coups pour la détruire
}

# ── Vérification ──────────────────────────────────────────
erreurs = []
for cle in ["x", "y", "couleur", "points", "resistance"]:
    if str(ma_brique.get(cle, '')) == '___' or ma_brique.get(cle) is None:
        erreurs.append(f'La clé "{cle}" est vide ou manquante')

if erreurs:
    for e in erreurs: print(f'  ❌ {e}')
else:
    print("✅ Brique créée avec succès !")
    for cle, val in ma_brique.items():
        print(f'  ' + 'ma_brique["' + cle + '"] = ' + repr(val))
````

## 🟢 Étape 2 — La fonction : une usine à briques

### 📖 Rappel
Une **fonction** est un bloc de code réutilisable. On lui donne des **paramètres**
et elle **retourne** un résultat.

```python
def ma_fonction(parametre1, parametre2):
    resultat = parametre1 + parametre2
    return resultat          # ← renvoie la valeur

ma_fonction(3, 5)  # → 8
```

### 🔍 Exemple : une fonction qui crée une brique
```python
def creer_brique(x, y, couleur, points, resistance):
    brique = {
        "x":          x,
        "y":          y,
        "w":          80,
        "h":          20,
        "couleur":    couleur,
        "points":     points,
        "resistance": resistance,
    }
    return brique

# Utilisation :
b = creer_brique(50, 80, "red", 10, 1)
print(b["couleur"])  # → "red"
```

### ✏️ Ton défi — Exercice 2
**Écris** la fonction `creer_brique()` en t'inspirant de l'exemple ci-dessus.
Attention : tu dois écrire **tout le corps** de la fonction, pas juste compléter des trous !

````python
# ════════════════════════════════════════════════════════════
#  EXERCICE 2 — Écris la fonction creer_brique()
#  Elle prend 5 paramètres et retourne un dictionnaire
# ════════════════════════════════════════════════════════════

def creer_brique(x, y, couleur, points, resistance):
    brique = {
        "x":          x,
        "y":          y,
        "w":          80,
        "h":          20,
        "couleur":    couleur,
        "points":     points,
        "resistance": resistance,
    }
    return brique
# ── Tests automatiques ────────────────────────────────────
# Ces lignes testent si ta fonction marche bien
try:
    b = creer_brique(100, 60, "blue", 20, 1)
    assert b is not None,            "La fonction doit retourner quelque chose (return !)"
    assert b["x"] == 100,           "x doit valoir 100"
    assert b["y"] == 60,            "y doit valoir 60"
    assert b["couleur"] == "blue",  "couleur doit valoir blue"
    assert b["points"] == 20,       "points doit valoir 20"
    assert b["resistance"] == 1,    "resistance doit valoir 1"
    assert b["w"] == 80,            "w doit valoir 80"
    assert b["h"] == 20,            "h doit valoir 20"
    print("✅ Parfait ! Ta fonction creer_brique() fonctionne !")
    print(f'   Elle retourne : {b}')
except AssertionError as e:
    print(f"❌ Presque ! {e}")
except TypeError as e:
    print(f"❌ Erreur dans la fonction : {e}")
````

## 🔵 Étape 3 — La boucle `for` : générer les rangées

### 📖 Rappel
Une **boucle `for`** répète du code pour chaque élément d'une liste.
```python
for i in range(5):   # i vaut 0, 1, 2, 3, 4
    x = 10 + i * 90  # x vaut 10, 100, 190, 280, 370
```

---

### 🗺️ Comprendre X et Y dans le jeu

L'écran du jeu fait **480 pixels de large** et **360 pixels de haut**.
`x` = position **horizontale** (de gauche à droite), `y` = position **verticale** (de haut en bas) :

```
x=0                    x=240                  x=480
 │                        │                      │
 ▼                        ▼                      ▼
┌──────────────────────────────────────────────────┐  ← y=0  (haut)
│                                                  │
│  [BRIQUE x=10]  [BRIQUE x=100]  [BRIQUE x=190]   │  ← y=50  (rangée 1)
│  [BRIQUE x=10]  [BRIQUE x=100]  [BRIQUE x=190]   │  ← y=80  (rangée 2)
│  [BRIQUE x=10]  [BRIQUE x=100]  [BRIQUE x=190]   │  ← y=110 (rangée 3)
│                                                  │
│                                                  │
│                    ●  ← balle                    │
│                                                  │
│              [══ RAQUETTE ══]                    │
└──────────────────────────────────────────────────┘  ← y=360 (bas)
```

**Règles à retenir :**
- Plus `x` est grand → la brique est plus à **droite**
- Plus `y` est grand → la brique est plus **bas**
- Les briques font **80px de large** → pour qu'elles ne se chevauchent pas, espace-les d'au moins 90px
- Les rangées sont séparées de **30px** (y=50, y=80, y=110...)

---

### 🔍 Exemple complet : une rangée de 5 briques

```python
#        i=0      i=1      i=2      i=3      i=4
#        ↓        ↓        ↓        ↓        ↓
# x =   10,      100,     190,     280,     370
# y =   50       50       50       50       50   ← même ligne !

for i in range(5):                        # 5 fois (i = 0, 1, 2, 3, 4)
    x = 10 + i * 90                       # positions : 10, 100, 190, 280, 370
    b = creer_brique(x, 50, "red", 10, 1) # x, y, couleur, points, resistance
    toutes_les_briques.append(b)           # on ajoute la brique à la liste
```

Pour faire une **2e rangée en dessous**, il suffit de changer `y` :
```python
for i in range(5):
    x = 10 + i * 90
    b = creer_brique(x, 80, "blue", 20, 1)  # ← y=80 au lieu de y=50 !
    toutes_les_briques.append(b)
```

### ✏️ Ton défi — Exercice 3
Crée **3 rangées** avec des couleurs, points et résistances différents.

````python
# ════════════════════════════════════════════════════════════
#  EXERCICE 3 — Génère 3 rangées de briques avec des boucles
# ════════════════════════════════════════════════════════════
#
#  La fonction creer_brique() s'utilise comme ça :
#    creer_brique(x, y, couleur, points, resistance)
#

toutes_les_briques = []   # liste qui contiendra TOUTES les briques

# ── Rangée 1 : y=50, briques faciles (1 coup, peu de points) ──────────────
# ✏️ Complète les ___ en t'aidant du schéma ci-dessus
for i in range(3):                              # ← combien de briques ? (5 c'est bien !)
    x = 10 + i * 90                               # x = 10, 100, 190, 280, 370
    b = creer_brique(x, 50, "yellow", 10, 1)          # ← couleur, points (ex:10), resistance (1)
    toutes_les_briques.append(b)

# ── Rangée 2 : y=80, briques moyennes (1 coup, plus de points) ────────────
# ✏️ Écris la boucle toi-même — même structure que la rangée 1, mais :
#    → change y (utilise 80)
#    → change la couleur
#    → mets plus de points (ex: 20)
for i in range(3):                              # ← combien de briques ? (5 c'est bien !)
    x = 10 + i * 90                               # x = 10, 100, 190, 280, 370
    b = creer_brique(x, 80, "red", 20, 1)        # ← couleur, points (ex:10), resistance (1)
    toutes_les_briques.append(b)




# ── Rangée 3 : y=110, briques solides (2 coups, beaucoup de points) ───────
# ✏️ Écris la boucle — cette fois resistance=2 (il faut 2 coups pour détruire !)
#    → change y (utilise 110)
#    → change la couleur
#    → mets encore plus de points (ex: 30)
#    → resistance = 2
for i in range(4):                              # ← combien de briques ? (5 c'est bien !)
    x = 10 + i * 90                           # x = 10, 100, 190, 280, 370
    b = creer_brique(x, 110, "blue", 50, 2)        # ← couleur, points (ex:10), resistance (1)
    toutes_les_briques.append(b)




# ── Vérification ────────────────────────────────────────────────
print(f'Nombre total de briques : {len(toutes_les_briques)}')
if len(toutes_les_briques) > 0:
    print(f'Première brique : x={toutes_les_briques[0]["x"]}, y={toutes_les_briques[0]["y"]}, couleur={toutes_les_briques[0]["couleur"]}')
    print(f'Dernière brique : x={toutes_les_briques[-1]["x"]}, y={toutes_les_briques[-1]["y"]}, résistance={toutes_les_briques[-1]["resistance"]}')
if len(toutes_les_briques) >= 10:
    print('✅ Super ! Tu as assez de briques pour remplir le jeu !')
else:
    print("⚠️ Essaie d'avoir au moins 10 briques au total (3 rangées de 5 = 15)")
````
