## 🔴 Étape 5 — Les briques spéciales 💣⭐

### 📖 Ce qu'on veut faire

### 🔍 Exemple : modifier des éléments au hasard dans une liste
```python
import random

def ajouter_etoiles(liste_mots, nb_etoiles):
    """Ajoute '⭐' devant nb_etoiles mots choisis au hasard."""
    indices_choisis = random.sample(range(len(liste_mots)), nb_etoiles)
    for i in indices_choisis:
        liste_mots[i] = "⭐ " + liste_mots[i]
    return liste_mots

mots = ["chat", "chien", "lapin", "souris", "hamster"]
ajouter_etoiles(mots, 2)
# → ["chat", "⭐ chien", "lapin", "⭐ souris", "hamster"]  (positions aléatoires)
```

> 💡 `random.sample(range(n), k)` choisit **k nombres différents** entre 0 et n-1

- Choisit `nb_bonus` autres briques au hasard et change leur `type_brique` en `"bonus"` et leur `couleur` en `"#FFD600"`
- Retourne la liste modifiée

> 💡 Pour modifier une brique : `briques[i]["type_brique"] = "bombe"`  
> 💡 Ne pas choisir les briques `"indestructible"` → filtre avec `if b["type_brique"] == "normale"`


````python
# ════════════════════════════════════════════════════════════
#  EXERCICE 4 — Écris ajouter_briques_speciales()
# ════════════════════════════════════════════════════════════
#
#  Rappel de l'exemple :
#    indices_choisis = random.sample(range(len(liste)), nb)
#    for i in indices_choisis:
#        liste[i]["champ"] = nouvelle_valeur
#
#  Étapes :
#    1. Trouve les indices des briques NORMALES
#       (boucle for avec enumerate, si type_brique == "normale")
#    2. Choisis nb_bombes indices au hasard parmi eux → random.sample(...)
#    3. Pour chaque indice choisi : change type_brique="bombe" et couleur="#FF4500"
#    4. Retire ces indices de la liste disponible
#    5. Même chose pour nb_bonus (type_brique="bonus", couleur="#FFD600")
#    6. return briques
# ════════════════════════════════════════════════════════════

import random

def ajouter_briques_speciales(briques, nb_bombes=2, nb_bonus=2):
    """Remplace aléatoirement des briques normales par des bombes et des bonus."""
    # ✏️ Étape 1 : trouve les indices des briques normales
    indices_normaux = []
    for i, b in enumerate(briques):
        if b["type_brique"] == "normale":
            indices_normaux.append(i)

    # ✏️ Étape 2 : choisit nb_bombes indices au hasard
    # nb_bombes = min(nb_bombes, len(indices_normaux)) pour ne pas dépasser
    nb_bombes = min(nb_bombes, len(indices_normaux))
    indices_bombes = random.sample(___, nb_bombes)   # ✏️ quelle liste donner à random.sample ?

    # ✏️ Étape 3 : transforme ces briques en bombes
    for i in indices_bombes:
        briques[i]["type_brique"] = "bombe"
        briques[i]["couleur"]     = "#FF4500"

    # ✏️ Étape 4 : retire les indices bombes de la liste disponible
    indices_restants = [i for i in indices_normaux if i not in indices_bombes]

    # ✏️ Étape 5 : même chose pour les bonus (couleur="#FFD600", type_brique="bonus")
    nb_bonus = min(nb_bonus, len(indices_restants))
    indices_bonus = random.sample(indices_restants, nb_bonus)
    for i in indices_bonus:
        briques[i]["type_brique"] = ___   # ✏️
        briques[i]["couleur"]     = ___   # ✏️

    return briques

# ── Tests automatiques ────────────────────────────────────
try:
    import copy
    test_briques = generer_niveau(1)
    original_len = len(test_briques)
    resultat = ajouter_briques_speciales(copy.deepcopy(test_briques), nb_bombes=2, nb_bonus=2)
    bombes = [b for b in resultat if b["type_brique"] == "bombe"]
    bonus  = [b for b in resultat if b["type_brique"] == "bonus"]
    assert len(bombes) == 2,  f"❌ Doit avoir 2 bombes, obtenu: {len(bombes)}"
    assert len(bonus)  == 2,  f"❌ Doit avoir 2 bonus, obtenu: {len(bonus)}"
    assert len(resultat) == original_len, "❌ La liste ne doit pas changer de taille"
    assert bombes[0]["couleur"] == "#FF4500", "❌ Couleur bombe doit être #FF4500"
    assert bonus[0]["couleur"]  == "#FFD600", "❌ Couleur bonus doit être #FFD600"
    print("✅ ajouter_briques_speciales() fonctionne !")
    print(f"   💣 {len(bombes)} bombes ajoutées")
    print(f"   ⭐ {len(bonus)} bonus ajoutés")
except AssertionError as e:
    print(f"❌ Presque ! {e}")
except TypeError as e:
    print(f"❌ Erreur : {e}")
````

````python
from IPython.display import display, HTML
import json

# ════════════════════════════════════════════════════════════
#  🔒 Moteur du jeu — ne pas modifier
# ════════════════════════════════════════════════════════════
try:
    assert len(briques) > 0
    assert couleur_raquette != "___"
except:
    print("⚠️ Exécute d'abord l'exercice 5 !")
else:
    config = {
        'vitesse':          vitesse,
        'raquette_largeur': 80,
        'couleur_raquette': couleur_raquette,
        'briques':          briques,
        'niveau':           numero_niveau,
    }
    cfg_json = json.dumps(config)

    GAME_HTML = """<!DOCTYPE html>
<html lang="fr">
<head>
<meta charset="UTF-8">
<style>
  * { margin:0; padding:0; box-sizing:border-box; }
  body { background:#0f0f1a; display:flex; flex-direction:column;
         align-items:center; font-family:'Segoe UI',sans-serif; color:#fff; padding:10px; }
  #wrapper { position:relative; }
  canvas { display:block; border:2px solid #333; border-radius:8px; background:#1a1a2e; }
  #ui { display:flex; justify-content:space-between; align-items:center;
        padding:8px 4px; width:100%; }
  .uv { font-size:16px; font-weight:700; color:#FFD600; }
  .ul { font-size:11px; color:#888; text-transform:uppercase; letter-spacing:1px; }
  .ub { text-align:center; }
  #msg { position:absolute; top:50%; left:50%; transform:translate(-50%,-50%);
         background:rgba(0,0,0,0.88); border:2px solid #FFD600; border-radius:14px;
         padding:24px 36px; text-align:center; display:none; z-index:10; }
  #msg h2 { font-size:22px; color:#FFD600; margin-bottom:8px; }
  #msg p  { font-size:14px; color:#ccc; margin-bottom:16px; }
  #btn-r { background:linear-gradient(135deg,#FFD600,#ff8f00); color:#1a1200;
            border:none; border-radius:8px; padding:10px 24px; font-size:14px;
            font-weight:900; cursor:pointer; }
  #btn-r:hover { opacity:0.85; }
  #toolbar { display:flex; justify-content:center; margin-top:8px; gap:10px; }
  .tbtn { background:linear-gradient(135deg,#FFD600,#ff8f00); color:#1a1200;
          border:none; border-radius:8px; padding:9px 20px; font-size:13px;
          font-weight:900; cursor:pointer; }
  .tbtn:hover { opacity:0.85; }
  #legend { margin-top:8px; display:flex; gap:16px; font-size:11px; color:#aaa; }
  .leg { display:flex; align-items:center; gap:4px; }
  .dot { width:12px; height:12px; border-radius:3px; }
</style>
</head>
<body>
<div id="wrapper">
  <div id="ui">
    <div class="ub"><div class="uv" id="s">0</div><div class="ul">Score</div></div>
    <div class="ub"><div class="uv" id="niv">1</div><div class="ul">Niveau</div></div>
    <div class="ub"><div class="uv" id="b">0</div><div class="ul">Briques</div></div>
    <div class="ub"><div class="uv" id="v">3</div><div class="ul">Vies</div></div>
  </div>
  <canvas id="c" width="480" height="360"></canvas>
  <div id="msg">
    <h2 id="mt">Gagné !</h2>
    <p  id="mp">Bravo !</p>
    <button id="btn-r" onclick="init()">🔄 Rejouer</button>
  </div>
</div>
<div id="toolbar">
  <button class="tbtn" onclick="init()">🔄 Recommencer</button>
</div>
<div id="legend">
  <div class="leg"><div class="dot" style="background:#FF4500"></div> Bombe 💣</div>
  <div class="leg"><div class="dot" style="background:#FFD600"></div> Bonus ⭐ (×2 pts)</div>
  <div class="leg"><div class="dot" style="background:#555"></div> Indestructible 🔒</div>
</div>
<script>
const C = __CFG__;
const W=480, H=360;
const cv = document.getElementById('c');
const ctx = cv.getContext('2d');
let bx,by,vx,vy,rx,briques,score,vies,run,aid,floaters=[];
const RH=12, BR=8;
const RW = C.raquette_largeur || 80;

function init(){
  score=0; vies=3; run=true; floaters=[];
  document.getElementById('msg').style.display='none';
  document.getElementById('niv').textContent = C.niveau||1;
  bx=W/2; by=H-80;
  vx=C.vitesse; vy=-C.vitesse;
  rx=(W-RW)/2;
  briques = C.briques.map(b=>{
    const copy = Object.assign({}, b);
    copy.resistance_max = b.resistance; // sauvegarde la résistance initiale
    return copy;
  });
  maj();
  if(aid) cancelAnimationFrame(aid);
  loop();
}

function loop(){
  if(!run) return;
  ctx.clearRect(0,0,W,H);
  move();
  dessiner_briques();
  dessiner_raquette();
  dessiner_balle();
  check_win();
  dessiner_floaters();
  maj();
  aid = requestAnimationFrame(loop);
}

function move(){
  bx+=vx; by+=vy;
  if(bx+BR>=W) vx=-Math.abs(vx);
  if(bx-BR<=0) vx= Math.abs(vx);
  if(by-BR<=0) vy= Math.abs(vy);
  // Raquette
  if(by+BR>=H-RH-10 && by+BR<=H-5 && bx>=rx && bx<=rx+RW){
    vy=-Math.abs(vy);
    let h=(bx-rx)/RW;
    vx=C.vitesse*(h<0.5?-1:1)*(0.5+Math.abs(h-0.5)*1.5);
  }
  // Balle perdue
  if(by>H){
    vies--;
    if(vies<=0) fin(false);
    else { bx=W/2; by=H-80; vx=C.vitesse; vy=-C.vitesse; }
  }
  // Collisions briques
  briques.forEach((b,idx)=>{
    if(b.resistance<=0) return;  // brique déjà détruite → skip
    // Test de collision (commun à tous les types y compris indestructible)
    if(bx+BR>b.x && bx-BR<b.x+b.w && by+BR>b.y && by-BR<b.y+b.h){
      // Calcule l'overlap sur chaque axe pour détecter la bonne face
      const overlapX = Math.min(bx+BR - b.x, b.x+b.w - (bx-BR));
      const overlapY = Math.min(by+BR - b.y, b.y+b.h - (by-BR));
      if(overlapX < overlapY){
        vx = -vx;  // touche par le côté gauche ou droit
      } else {
        vy = -vy;  // touche par le haut ou le bas
      }
      // Briques indestructibles : rebond uniquement, pas de dégâts
      if(b.type_brique==='indestructible') return;
      b.resistance--;
      if(b.resistance<=0){
        let pts = b.points;
        if(b.type_brique==='bonus'){
          pts *= 2;
          // Affiche un message flottant ⭐ BONUS x2
          floaters.push({x:b.x+b.w/2, y:b.y, txt:'⭐ x2 +'+pts, color:'#FFD600', life:60});
        } else {
          floaters.push({x:b.x+b.w/2, y:b.y, txt:'+'+pts, color:'#fff', life:40});
        }
        score += pts;
      }
      if(b.type_brique==='bombe' && b.resistance<=0){
        // Explose les voisins
        briques.forEach(v=>{
          if(v!==b && v.resistance>0 && v.type_brique!=='indestructible'){
            const dx=Math.abs((v.x+v.w/2)-(b.x+b.w/2));
            const dy=Math.abs((v.y+v.h/2)-(b.y+b.h/2));
            if(dx<100 && dy<50){
              v.resistance=0;
              score+=v.points;
              floaters.push({x:v.x+v.w/2, y:v.y, txt:'💥 +'+v.points, color:'#FF6B6B', life:40});
            }
          }
        });
      }
      if(b.resistance>0){
        // Assombrit progressivement selon les dégâts reçus
        const pct = b.resistance / (b.resistance_max || b.resistance + 1);
        b.couleur = assombrir_pct(b.couleur_base||b.couleur, pct);
      }
    }
  });
}

function assombrir_pct(hex, pct){
  // pct = 1.0 → couleur normale, 0.0 → très sombre
  // pct = resistance_restante / resistance_max
  try{
    let c=hex.replace('#','');
    if(c.length===3) c=c[0]+c[0]+c[1]+c[1]+c[2]+c[2];
    let r=parseInt(c.slice(0,2),16),g=parseInt(c.slice(2,4),16),b=parseInt(c.slice(4,6),16);
    // Mélange la couleur avec du noir selon pct
    const factor = 0.3 + pct * 0.7; // va de 0.3 (sombre) à 1.0 (pleine couleur)
    r=Math.round(r*factor);
    g=Math.round(g*factor);
    b=Math.round(b*factor);
    return `rgb(${r},${g},${b})`;
  }catch(e){return hex;}
}

function dessiner_balle(){
  ctx.beginPath();
  ctx.arc(bx,by,BR,0,Math.PI*2);
  ctx.fillStyle='white';
  ctx.fill();
}

function dessiner_raquette(){
  ctx.fillStyle=C.couleur_raquette;
  ctx.beginPath();
  ctx.roundRect(rx,H-RH-10,RW,RH,4);
  ctx.fill();
}

function dessiner_briques(){
  briques.forEach(b=>{
    if(b.resistance<=0) return;
    ctx.fillStyle=b.couleur;
    ctx.beginPath();
    ctx.roundRect(b.x,b.y,b.w,b.h,3);
    ctx.fill();
    ctx.fillStyle='rgba(255,255,255,0.18)';
    ctx.fillRect(b.x+3,b.y+3,b.w-6,4);
    // Icone type
    let ico='';
    if(b.type_brique==='bombe') ico='💣';
    else if(b.type_brique==='bonus') ico='⭐';
    else if(b.type_brique==='indestructible') ico='🔒';
    ctx.fillStyle='rgba(255,255,255,0.8)';
    ctx.font='bold 9px Segoe UI';
    ctx.textAlign='center';
    if(ico){
      ctx.font='10px Segoe UI';
      ctx.fillText(ico,b.x+b.w/2,b.y+b.h-4);
    } else {
      ctx.fillText(b.points+'pts',b.x+b.w/2,b.y+b.h-5);
    }
    if(b.resistance_max>1 && b.type_brique!=='indestructible'){
      // Barre de vie mini en haut de la brique
      const bw = b.w - 6;
      const pct_hp = b.resistance / b.resistance_max;
      // Fond gris
      ctx.fillStyle='rgba(0,0,0,0.5)';
      ctx.fillRect(b.x+3, b.y+3, bw, 4);
      // Barre colorée selon PV restants
      const bar_color = pct_hp>0.6?'#4CAF50':pct_hp>0.3?'#FF9800':'#f44336';
      ctx.fillStyle=bar_color;
      ctx.fillRect(b.x+3, b.y+3, Math.round(bw*pct_hp), 4);
    }
  });
}

function dessiner_floaters(){
  floaters = floaters.filter(f=>f.life>0);
  floaters.forEach(f=>{
    ctx.save();
    ctx.globalAlpha = Math.min(1, f.life/20);
    ctx.font = 'bold 14px Segoe UI';
    ctx.fillStyle = f.color;
    ctx.textAlign = 'center';
    ctx.shadowColor = 'rgba(0,0,0,0.8)';
    ctx.shadowBlur = 4;
    ctx.fillText(f.txt, f.x, f.y);
    ctx.restore();
    f.y -= 1.5;   // monte vers le haut
    f.life--;
  });
}

function check_win(){
  const actives = briques.filter(b=>b.resistance>0&&b.type_brique!=='indestructible');
  if(actives.length===0) fin(true);
}

function fin(w){
  run=false;
  document.getElementById('mt').textContent=w?'🏆 Gagné !':'💀 Perdu !';
  document.getElementById('mp').textContent=w?`Score final : ${score} pts !`:`${score} pts — réessaie !`;
  document.getElementById('msg').style.display='block';
}

function maj(){
  document.getElementById('s').textContent=score;
  const actives=briques.filter(b=>b.resistance>0&&b.type_brique!=='indestructible');
  document.getElementById('b').textContent=actives.length;
  document.getElementById('v').textContent=vies;
}

let keys={};
document.addEventListener('keydown',e=>{keys[e.key]=true;e.preventDefault();});
document.addEventListener('keyup',e=>{keys[e.key]=false;});
cv.addEventListener('mousemove',e=>{
  const r=cv.getBoundingClientRect();
  rx=Math.max(0,Math.min(W-RW,e.clientX-r.left-RW/2));
});
setInterval(()=>{
  if(!run)return;
  if(keys['ArrowLeft'])  rx=Math.max(0,rx-6);
  if(keys['ArrowRight']) rx=Math.min(W-RW,rx+6);
},16);
init();
</script>
</body>
</html>"""
    final = GAME_HTML.replace('__CFG__', cfg_json)
    display(HTML(f'<div style="max-width:520px;margin:0 auto;">{final}</div>'))
````
