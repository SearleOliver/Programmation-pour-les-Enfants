Maintenant il faut pouvoir placer un bateau dans cette grille !

 - La fonction placer_bateau
 
Placer_bateau est déjà écrit , tu dois compléter verifier placement pour la faire fonctionner !

Pour la simplicité **orientation** sera soit Horizontale ("H"), soit verticale ("V").

tu vas devoir vérifier deux choses 
1. Est ce que ton bateau n'est pas placé sur un autre
2. Est ce que ton bateau dépasse du plateau (en hauteur ou en largeur)

Complète la fonction verifier_emplacement  :

```python
def verifier_placement(grille, ligne, colonne, taille, orientation):
   
    n = len(grille)

    if orientation == "H":
        # Sortie de grille
        """
        """
        if colonne + taille > n:
            return False

        # Collision
        """
        """
        for i in range(taille):
            if grille[ligne][colonne + i] != "0":
                return False

    elif orientation == "V":
        # Sortie de grille
        if ligne + taille > n:
            return False

        # Collision
        for i in range(taille):
            if grille[ligne + i][colonne] != "0":
                return False
    
    return True  
    
    

def placer_bateau(grille, ligne, colonne, taille, orientation):
    if verifier_placement(grille, ligne, colonne, taille, orientation) :
        if orientation == "H":
            for i in range(taille):
                grille[ligne][colonne + i] = VERT+ "X" +RESET
            return True
        else:  
            for i in range(taille):
                grille[ligne + i][colonne] = VERT +"X"+ RESET
            return True
    return False


```

    Exercice 6 - Boucle de jeu 

On arrive à la dernière partie de ce mini projet , il faut créer une partie !

Il faut initialiser 4 grilles, 2 pour chaque joueur l'une sera sa grille personnelle et l'autre sera sa grille afficher .
Ensuite chaque joueur doit placer ses bateaux .
La prochaine étape est de faire une boucle jusqu'a la victoire d'un des deux joueurs !
Les étape seront les suivantes Tire du joueur 1 puis tire du joueur 2 , si toutes les cases contenant des bateaux sont toucher alors 

```python
def partie():
    
    grille_j1 = creer_grille(10)          
    grille_affichage_j1 = creer_grille(10)  

    
    grille_j2 = creer_grille(10)
    grille_affichage_j2 = creer_grille(10)

    print("\n==============================")
    print(" Placement des bateaux J1 ")
    print("==============================")

    placement_joueur(grille_j1)

    print("\n==============================")
    print(" Placement des bateaux J2 ")
    print("==============================")

    placement_joueur(grille_j2)



    total_cases = 0

    for taille in bateaux.values():
        total_cases += taille


    pv_j1 = total_cases
    pv_j2 = total_cases



    victoire = False



    while victoire == False:
        print("\n==============================", flush=True)
        print(" Tour du Joueur 1 ", flush=True)
        print("==============================", flush=True)

        resultat = tire(grille_j2, grille_affichage_j2, grille_affichage_j2)

        if resultat == True:
            pv_j2 -= 1

        print(f"Cases restantes J2 : {pv_j2}", flush=True)

        if pv_j2 == 0:
            print("\n🏆 Le Joueur 1 a gagné !", flush=True)
            victoire = True
            break


        print("\n==============================")
        print(" Tour du Joueur 2 ")
        print("==============================")

        resultat = tire(grille_j1, grille_affichage_j1, grille_affichage_j1)

        if resultat == False:
            pv_j1 -= 1

        print(f"Cases restantes J1 : {pv_j1}", flush=True)

        if pv_j1 == 0:
            print("\n🏆 Le Joueur 2 a gagné !", flush=True)
            victoire = True
            break


partie()
```
