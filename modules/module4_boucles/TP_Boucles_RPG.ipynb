# 🏰 Quête 1 — L'entraînement du héros
### 📖 Concept : Boucle `for` avec `range()`

---

## 📖 Rappel

```python
for i in range(5):
    print("Le héros s'entraîne !")
# → affiche 5 fois la même ligne
```

> 💡 `range(n)` génère les nombres de **0 à n-1**
> 💡 `range(1, 6)` génère les nombres de **1 à 5**
> 💡 `range(0, 10, 2)` compte de 0 à 9 **par pas de 2**


### ⚔️ Échauffement — Exécute et observe !

````python
# Exécute pour voir range() en action !
print("=== range(4) ===")
for i in range(4):
    print("Tour d'entraînement n°", i)

print()
print("=== range(1, 6) ===")
for niveau in range(1, 6):
    print("Niveau", niveau, "atteint !")

print()
print("=== range(0, 50, 10) ===")
for xp in range(0, 50, 10):
    print("XP :", xp)
````

# 🎒 Quête 2 — L'inventaire du guerrier
### 📖 Concept : Boucle `for` sur une liste

---

## 📖 Rappel

```python
inventaire = ["Épée", "Bouclier", "Potion"]

for objet in inventaire:
    print("Tu possèdes :", objet)
```

> 💡 À chaque tour, la variable `objet` prend la valeur de l'élément suivant.


### ⚔️ Échauffement — Exécute et observe !

````python
inventaire = ["Épée de feu", "Bouclier magique", "Potion de vie", "Arc elfique"]

print("=== Inventaire simple ===")
for objet in inventaire:
    print("🎒", objet)

print()
print("=== Avec enumerate (numéro + élément) ===")
for i, objet in enumerate(inventaire):
    print(str(i+1) + ".", objet)
````

# 🐉 Quête 3 — Le combat contre le boss
### 📖 Concept : Boucle `while`

---

## 📖 Rappel

```python
pv_boss = 50

while pv_boss > 0:
    print("Le boss a encore", pv_boss, "PV !")
    pv_boss = pv_boss - 15

print("Le boss est vaincu !")
```

> ⚠️ Il faut **toujours modifier** la variable testée dans la boucle,
> sinon elle tourne à l'infini !


### ⚔️ Échauffement — Exécute et observe !

````python
# Exécute pour voir while en action !
pv_hero = 30

print("🗡️ Combat en cours...")
while pv_hero > 0:
    print("  Le héros a " + str(pv_hero) + " PV — ", end="")
    pv_hero = pv_hero - 8
    if pv_hero > 0:
        print("il résiste !")
    else:
        print("K.O. !")

print("💀 Le héros est tombé au combat !")
````

### 🎯 Manche 7 — Les niveaux du donjon 🗺️

**Mission :** Le donjon a **3 niveaux**. À chaque niveau, un monstre différent attend le héros.
Le héros combat chaque monstre jusqu'à ce que l'un des deux soit K.O.

Structure :
```
for chaque monstre dans la liste:     ← for (nombre de niveaux fixe)
    while le héros et le monstre sont vivants:  ← while (combat)
        un tour de combat
```


````python
import random

# Les monstres du donjon
monstres = [
    {"nom": "Gobelin",    "pv": 20, "attaque": 6},
    {"nom": "Troll",      "pv": 35, "attaque": 10},
    {"nom": "Dragon Boss","pv": 50, "attaque": 15},
]

# ✏️ MANCHE 7 — Complète les ___ pour le donjon !

victoires = 0

for monstre in ___:                             # ✏️ parcours les monstres
    pv_hero    = 40
    pv_monstre = monstre["___"]                 # ✏️ clé pour les PV

    print(f"\n⚔️  Niveau {victoires+1} : Héros VS {monstre['nom']}")
    print("─" * 35)
    tour = 1

    while pv_hero > 0 and ___ > 0:             # ✏️ les deux encore vivants

        # Le héros attaque (12 dégâts ± hasard)
        degats_hero   = 12 + random.randint(-2, 2)
        pv_monstre    = max(0, pv_monstre - degats_hero)

        # Le monstre riposte
        degats_monstre = monstre["attaque"] + random.randint(-1, 1)
        pv_hero        = max(0, pv_hero - degats_monstre)

        print(f"  Tour {tour} | Héros: {pv_hero} PV | {monstre['nom']}: {pv_monstre} PV")
        tour += ___                             # ✏️ incrémente le tour

    # Résultat du niveau
    if ___ > 0:                                 # ✏️ le héros a encore des PV ?
        print(f"✅ Victoire ! Le héros bat {monstre['nom']} !")
        victoires += 1
    else:
        print(f"💀 Défaite face au {monstre['nom']}…")

print(f"\n🏆 Donjon terminé : {victoires}/{len(monstres)} niveaux complétés !")
````
