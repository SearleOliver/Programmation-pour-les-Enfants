# 🌿 BADGE FORÊT — La boucle `for` avec `range()`
### 📖 Répéter un nombre de fois fixe
---

## 📖 Rappel

```python
for i in range(4):
    print("Pikachu attaque !")
# → affiche 4 fois "Pikachu attaque !"
```

> 💡 `range(n)` génère les nombres de **0 à n-1**
> 💡 `range(1, 6)` génère les nombres de **1 à 5**


### ⚡ Échauffement — Exécute et observe !

````python
# Exécute pour voir range() en action !
print("=== range(5) ===")
for i in range(5):
    print("Tour", i)

print()
print("=== range(1, 4) ===")
for i in range(1, 4):
    print("Niveau", i)

print()
print("=== range(0, 10, 2) ===")  # de 0 à 9, par pas de 2
for i in range(0, 10, 2):
    print(i, end=" ")
print()
````

# 🔥 BADGE FLAMME — La boucle `while`
### 📖 Répéter TANT QUE une condition est vraie
---

## 📖 Rappel

```python
pv = 30

while pv > 0:
    print("Pikachu a " + str(pv) + " PV")
    pv = pv - 10

# La boucle s'arrête quand pv <= 0
```

> ⚠️ Il faut **toujours modifier** la variable testée dans la boucle, sinon elle tourne à l'infini !


### ⚡ Échauffement — Exécute et observe !

````python
# Exécute pour voir while en action !
pv = 25

print("🏥 Combat en cours...")
while pv > 0:
    print("  Pikachu a " + str(pv) + " PV — ", end="")
    pv = pv - 7   # Ronflex inflige 7 dégâts
    if pv > 0:
        print("il tient bon !")
    else:
        print("K.O. !")

print("💀 Pikachu est K.O. !")
````

Pour chaque combat, Pikachu affronte un adversaire différent.
Chaque combat dure **jusqu'à ce que** l'un des deux soit K.O.

Structure :
```
for chaque adversaire dans la liste:     ← for (nombre de combats fixe)
    while les deux Pokémon sont vivants: ← while (on ne sait pas combien de tours)
        un tour de combat
```


````python
import random

# Les adversaires du tournoi
adversaires = [
    {"nom": "Bulbizarre",  "pv": 28, "attaque": 10},
    {"nom": "Ronflex",     "pv": 40, "attaque": 14},
    {"nom": "Dracaufeu",   "pv": 35, "attaque": 18},
]

# ✏️ DÉFI 7 — Complète les ___ pour le tournoi !

victoires = 0

for adversaire in ___:                          # ✏️ parcours les adversaires
    # Remet Pikachu à pleine forme avant chaque combat
    pv_pikachu   = 35
    pv_adversaire = adversaire["___"]           # ✏️ clé pour les PV

    print(f"\n⚔️  Combat : Pikachu VS {adversaire['nom']}")
    print("─" * 35)
    tour = 1

    while pv_pikachu > 0 and ___ > 0:          # ✏️ condition : les deux vivants

        # Pikachu attaque (15 dégâts ± hasard)
        degats_pikachu   = 15 + random.randint(-3, 3)
        pv_adversaire    = max(0, pv_adversaire - degats_pikachu)

        # L'adversaire riposte
        degats_adversaire = adversaire["attaque"] + random.randint(-2, 2)
        pv_pikachu        = max(0, pv_pikachu - degats_adversaire)

        print(f"  Tour {tour} | Pikachu: {pv_pikachu} PV | {adversaire['nom']}: {pv_adversaire} PV")
        tour += ___                             # ✏️ incrémente le tour

    # Résultat du combat
    if ___ > 0:                                 # ✏️ Pikachu a encore des PV ?
        print(f"✅ Victoire ! Pikachu bat {adversaire['nom']} !")
        victoires += 1
    else:
        print(f"💀 Défaite face à {adversaire['nom']}…")

print(f"\n🏆 Tournoi terminé : {victoires}/{len(adversaires)} victoires !")
````

### 🌟 Bonus 1 — Compte les Pokémon par type 📊

À partir d'une liste de Pokémon avec leurs types, compte combien il y en a de chaque type en utilisant une boucle `for` et un dictionnaire.


````python
# 🌟 BONUS 1 — Compte par type

pokedex = [
    {"nom": "Pikachu",    "type": "Électrik"},
    {"nom": "Salamèche",  "type": "Feu"},
    {"nom": "Bulbizarre", "type": "Plante"},
    {"nom": "Dracaufeu",  "type": "Feu"},
    {"nom": "Carapuce",   "type": "Eau"},
    {"nom": "Ponyta",     "type": "Feu"},
    {"nom": "Ronflex",    "type": "Normal"},
]

# ✏️ Compte le nombre de Pokémon par type
# Utilise un dictionnaire : {"Feu": 2, "Eau": 1, ...}
compte_types = {}

for pokemon in pokedex:
    type_pokemon = pokemon["type"]
    if type_pokemon in compte_types:
        compte_types[type_pokemon] = compte_types[type_pokemon] + 1
    else:
        compte_types[type_pokemon] = 1

print("📊 Pokémon par type :")
for type_name, nb in compte_types.items():
    barre = "█" * nb
    print(f"  {type_name:<12} {barre} ({nb})")
````

### 🌟 Bonus 2 — Le jeu de devinette Pokémon 🎲

Pikachu pense à un nombre entre 1 et 20.
Utilise une boucle `while` pour deviner en proposant des nombres aléatoires.
Affiche combien d'essais il a fallu !


````python
# 🌟 BONUS 2 — Devinette Pokémon

import random

secret = random.randint(1, 20)   # Le nombre secret de Pikachu
essai  = 0
trouve = False

print("⚡ Pikachu pense à un nombre entre 1 et 20...")
print("   (Simulation : l'ordi devine tout seul !)")

while not trouve:
    proposition = random.randint(1, 20)
    essai += 1

    if proposition < secret:
        print(f"  Essai {essai} : {proposition} → Trop petit !")
    elif proposition > secret:
        print(f"  Essai {essai} : {proposition} → Trop grand !")
    else:
        print(f"  Essai {essai} : {proposition} → 🎉 Trouvé !")
        trouve = True

print(f"\n✅ Trouvé en {essai} essais ! Le nombre était {secret}.")
````
